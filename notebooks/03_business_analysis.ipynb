5. How do trip distance and duration relate to fares?

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
ANALYSIS_DIR = PROJECT_ROOT / "reports" / "analysis"

monthly_demand = pd.read_csv(ANALYSIS_DIR / "monthly_demand.csv")
hourly_demand = pd.read_csv(ANALYSIS_DIR / "hourly_demand.csv")
day_type_demand = pd.read_csv(ANALYSIS_DIR / "day_type_demand.csv")
hour_day_demand = pd.read_csv(ANALYSIS_DIR / "hour_day_demand.csv")

top_pickup_zones = pd.read_csv(ANALYSIS_DIR / "top_pickup_zones.csv")
top_dropoff_zones = pd.read_csv(ANALYSIS_DIR / "top_dropoff_zones.csv")
borough_demand = pd.read_csv(ANALYSIS_DIR / "borough_demand.csv")

monthly_economics = pd.read_csv(ANALYSIS_DIR / "monthly_economics.csv")
payment_methods = pd.read_csv(ANALYSIS_DIR / "payment_methods.csv")

distance_economics = pd.read_csv(ANALYSIS_DIR / "distance_economics.csv")
duration_economics = pd.read_csv(ANALYSIS_DIR / "duration_economics.csv")
airport_economics = pd.read_csv(ANALYSIS_DIR / "airport_vs_non_airport.csv")

print("Analysis datasets loaded successfully.")
```

## 1. When Does Taxi Demand Peak?

### Finding

Taxi demand varies substantially across the day, with demand reaching its
highest hourly volume during the evening period.

The analysis identifies **6 PM (18:00)** as the highest-volume hour across
the 12-month study period.

The lowest hourly volume occurs at **4 AM (04:00)**.

### Operational Interpretation

The concentration of trips during the afternoon and evening suggests that
taxi activity is not evenly distributed throughout the day. Peak-period
planning should therefore account for substantially different demand levels
across operating hours.

These findings describe observed demand patterns and do not establish the
causes of those patterns.

```python
peak_hour = hourly_demand.loc[
    hourly_demand["total_trips"].idxmax()
]

lowest_hour = hourly_demand.loc[
    hourly_demand["total_trips"].idxmin()
]

print(
    f"Peak hour: {int(peak_hour['pickup_hour']):02d}:00 "
    f"— {int(peak_hour['total_trips']):,} trips"
)

print(
    f"Lowest hour: {int(lowest_hour['pickup_hour']):02d}:00 "
    f"— {int(lowest_hour['total_trips']):,} trips"
)
```

## 2. How Does Demand Vary by Month?

### Finding

Monthly taxi demand varies across the 12-month study period.

**December 2025** records the highest monthly trip volume, while **August
2025** records the lowest.

The monthly pattern indicates that demand changes over the year rather than
remaining constant.

### Key Values

- Highest: December 2025 — 4,256,070 trips
- Lowest: August 2025 — 3,312,598 trips

The analysis is descriptive and does not determine the causes of monthly
variation.

```python
highest_month = monthly_demand.loc[
    monthly_demand["total_trips"].idxmax()
]

lowest_month = monthly_demand.loc[
    monthly_demand["total_trips"].idxmin()
]

highest_month_name = pd.Timestamp(
    year=int(highest_month["pickup_year"]),
    month=int(highest_month["pickup_month"]),
    day=1
).strftime("%B %Y")

lowest_month_name = pd.Timestamp(
    year=int(lowest_month["pickup_year"]),
    month=int(lowest_month["pickup_month"]),
    day=1
).strftime("%B %Y")

print(
    f"Highest-demand month: "
    f"{highest_month_name} — "
    f"{int(highest_month['total_trips']):,} trips"
)

print(
    f"Lowest-demand month: "
    f"{lowest_month_name} — "
    f"{int(lowest_month['total_trips']):,} trips"
)
```

## 5. How Do Trip Distance and Duration Relate to Fares?

### Finding

Trip economics vary substantially with both distance and duration.

Average fare increases across most distance bands, with longer trips associated
with higher average fares and total trip amounts.

Trip duration shows a similar broad relationship with fare, although the
shortest duration bands are not perfectly monotonic.

These are descriptive relationships in the dataset and should not be
interpreted as evidence that distance or duration alone causes the fare change.

```python
print("Distance economics:")
display(distance_economics)

print("\nDuration economics:")
display(duration_economics)
```

```python
distance_highest = distance_economics.loc[
    distance_economics["avg_fare"].idxmax()
]

distance_lowest = distance_economics.loc[
    distance_economics["avg_fare"].idxmin()
]

print(
    f"Lowest average fare band: "
    f"{distance_lowest['distance_band']} — "
    f"${distance_lowest['avg_fare']:.2f}"
)

print(
    f"Highest average fare band: "
    f"{distance_highest['distance_band']} — "
    f"${distance_highest['avg_fare']:.2f}"
)
```

```python
duration_highest = duration_economics.loc[
    duration_economics["avg_fare"].idxmax()
]

duration_lowest = duration_economics.loc[
    duration_economics["avg_fare"].idxmin()
]

print(
    f"Lowest average fare duration band: "
    f"{duration_lowest['duration_band']} — "
    f"${duration_lowest['avg_fare']:.2f}"
)

print(
    f"Highest average fare duration band: "
    f"{duration_highest['duration_band']} — "
    f"${duration_highest['avg_fare']:.2f}"
)
```

## 6. How Does Trip Economics Vary Across Months?

### Finding

Monthly trip volume and trip economics do not move identically across the
study period.

December 2025 records the highest total trip amount, while August 2025 records
the lowest. Average trip amounts also vary between months, indicating that
changes in total trip value are influenced by both trip volume and the average
value of individual trips.

These results are descriptive and do not establish the causes of monthly
variation.

```python
print("Monthly trip economics:")
display(monthly_economics)
```

```python
highest_total = monthly_economics.loc[
    monthly_economics["total_amount"].idxmax()
]

lowest_total = monthly_economics.loc[
    monthly_economics["total_amount"].idxmin()
]

print(
    f"Highest total trip amount: "
    f"{int(highest_total['pickup_year'])}-"
    f"{int(highest_total['pickup_month']):02d} — "
    f"${highest_total['total_amount']:,.2f}"
)

print(
    f"Lowest total trip amount: "
    f"{int(lowest_total['pickup_year'])}-"
    f"{int(lowest_total['pickup_month']):02d} — "
    f"${lowest_total['total_amount']:,.2f}"
)
```

```python
highest_avg = monthly_economics.loc[
    monthly_economics["avg_total_amount"].idxmax()
]

lowest_avg = monthly_economics.loc[
    monthly_economics["avg_total_amount"].idxmin()
]

print(
    f"Highest average trip amount: "
    f"{int(highest_avg['pickup_year'])}-"
    f"{int(highest_avg['pickup_month']):02d} — "
    f"${highest_avg['avg_total_amount']:.2f}"
)

print(
    f"Lowest average trip amount: "
    f"{int(lowest_avg['pickup_year'])}-"
    f"{int(lowest_avg['pickup_month']):02d} — "
    f"${lowest_avg['avg_total_amount']:.2f}"
)
```

## 7. How Do Payment Methods Contribute to Trip Economics?

### Finding

Payment-type codes are unevenly distributed across the dataset. Payment type
code 1 represents the largest share of recorded trips, followed by codes 0
and 2.

The analysis also shows differences in average trip amount and average tip
across payment-type codes.

Because the TLC dataset stores payment methods as numeric codes, the codes
should be mapped to their official TLC definitions before presenting named
payment methods in the final dashboard.

This analysis describes the recorded payment-type distribution and does not
establish why the observed differences occur.

```python
print("Payment method analysis:")
display(payment_methods)
```

```python
largest_payment_type = payment_methods.loc[
    payment_methods["total_trips"].idxmax()
]

print(
    f"Largest payment-type code: "
    f"{int(largest_payment_type['payment_type'])}"
)

print(
    f"Trips: {int(largest_payment_type['total_trips']):,}"
)

print(
    f"Trip share: {largest_payment_type['trip_percentage']:.2f}%"
)
```

## 8. How Do Airport-Originating Trips Differ From Other Trips?

### Finding

Airport-originating trips account for a smaller share of total trips but have
substantially different trip characteristics from non-airport trips.

Airport-originating trips have higher average distance, duration, fare, and
total trip amount than non-airport trips.

Airport trips are defined here as trips whose pickup location is JFK Airport
or LaGuardia Airport. Therefore, this comparison concerns airport-originating
trips only and does not represent every trip involving an airport.

The results are descriptive and do not establish the causes of these
differences.

```python
print("Airport vs non-airport analysis:")
display(airport_economics)
```

```python
airport_trips = airport_economics.loc[
    airport_economics["trip_type"] == "Airport",
    "total_trips"
].iloc[0]

total_trips = airport_economics["total_trips"].sum()

airport_share = airport_trips / total_trips * 100

print(f"Airport-originating trips: {airport_trips:,}")
print(f"Airport trip share: {airport_share:.2f}%")
```
